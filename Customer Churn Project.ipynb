```python
#Import Pandas for creating, reading, and analyzing tabular data (DataFrames)- Data handling & analysis - Read Excel/CSV, filter data
import pandas as pd 
# Import NumPy for numerical calculations and working with arrays - Mathematical calculations - Arrays, averages, calculations
import numpy as np 
# Import Matplotlib for creating graphs and visualizations - Basic graphs - Bar chart, line graph, pie chart
import matplotlib.pyplot as plt 
# Import Seaborn for creating attractive and statistical data visualizations - Advanced/beautiful graphs -Heatmap, distribution plots, statistical graphs
import seaborn as sns 

```

```python
df = pd.read_csv('Customer Churn.csv')
df
```

```python
df.info()
```

```python
df['TotalCharges'] = df['TotalCharges'].replace(" ","0")
```

```python
df['TotalCharges'] = df['TotalCharges'].astype("float")
```

```python
def conv(value):
    if value==1:
        return 'yes'
    else:
        return 'no'
df['SeniorCitizen'] = df['SeniorCitizen'].apply(conv)
```

```python
df.head()
```

```python
ax = sns.countplot(x = 'Churn', data =df)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
plt.figure(figsize=(3,4))
gb = df.groupby("Churn").agg({'Churn':'count'})
plt.pie(gb['Churn'],labels=gb.index,autopct = "%1.2f%%")
plt.title("Percentage of churned Customers",fontsize=10)
plt.show()
```

```python
plt.figure(figsize=(4,4))
sns.countplot(x = 'gender',data=df, hue = 'Churn')
plt.title('Churn by Gender')
plt.show()
```

```python
sns.countplot(x='SeniorCitizen',data=df,hue = 'Churn')
plt.title("Churn by Senior citizen")
plt.show()
```

```python

# Calculate percentage distribution
percentage = pd.crosstab(
    df['SeniorCitizen'],
    df['Churn'],
    normalize='index'
) * 100

# Create stacked bar chart
ax = percentage.plot(
    kind='bar',
    stacked=True,
    figsize=(5, 3)
     
)

# Add percentage labels
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.1f%%',
        label_type='center'
    )

plt.title("Churn by Senior Citizen")
plt.xlabel("Senior Citizen")
plt.ylabel("Percentage")
plt.legend(title="Churn")
plt.ylim(0, 100)
plt.show()
```
